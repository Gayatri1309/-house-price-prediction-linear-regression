## Feature Selection

The target variable for this project is **Price_INR_Lakhs**, which represents the house price in Indian Rupees (Lakhs).

The following features are selected because they are expected to influence house prices:

- **Super_Area_SqFt** – Larger properties generally have higher prices.
- **BHK** – The number of bedrooms can affect property value.
- **Bathrooms** – More bathrooms may increase the property's value.
- **Age_of_Property** – Older properties may have lower prices compared to newer properties.
- **City** – Property prices can vary significantly across different cities.
- **Locality_Type** – Premium, Mid-Range, and Affordable localities can influence prices.
- **Property_Type** – Apartment, Villa, Independent House, etc. may have different price levels.
- **Furnishing_Status** – Furnished properties may have higher prices.
- **Distance_to_Metro_km** – Distance from metro stations can influence property value.
- **Distance_to_City_Center_km** – Properties closer to city centers may have higher prices.

The **Property_ID** column is not selected because it is only an identifier and does not provide meaningful information for predicting house prices.

The target variable is:

**Price_INR_Lakhs**

```python
print("Missing values before preprocessing:")

display(df.isnull().sum())
```

```python
features = [
    "Super_Area_SqFt",
    "BHK",
    "Bathrooms",
    "Age_of_Property",
    "City",
    "Locality_Type",
    "Property_Type",
    "Furnishing_Status",
    "Distance_to_Metro_km",
    "Distance_to_City_Center_km"
]

target = "Price_INR_Lakhs"

X = df[features]
y = df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=[
        "City",
        "Locality_Type",
        "Property_Type",
        "Furnishing_Status"
    ],
    drop_first=True
)

print("Shape after One-Hot Encoding:", X_encoded.shape)

display(X_encoded.head())
```

```python
correlation_matrix = df.select_dtypes(include="number").corr()

display(correlation_matrix)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of House Price Dataset")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
# Predict house prices for test data

y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print("First 10 Predicted Prices:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print("MSE  :", mse)
print("RMSE :", rmse)
print("R²   :", r2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.6)

plt.xlabel("Actual House Price (INR Lakhs)")
plt.ylabel("Predicted House Price (INR Lakhs)")
plt.title("Actual vs Predicted House Prices")

min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate residuals
residuals = y_test - y_pred

plt.figure(figsize=(8, 6))

plt.scatter(y_pred, residuals, alpha=0.6)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Price (INR Lakhs)")
plt.ylabel("Residuals")
plt.title("Residual Plot")

plt.show()
```

```python
# Create coefficient table

coefficients = pd.DataFrame({
    "Feature": X_encoded.columns,
    "Coefficient": model.coef_
})

# Determine positive or negative impact
coefficients["Impact"] = np.where(
    coefficients["Coefficient"] > 0,
    "Positive",
    "Negative"
)

# Sort by coefficient
coefficients = coefficients.sort_values(
    by="Coefficient",
    ascending=False
)

print("Feature Coefficient Analysis")
print("----------------------------")

display(coefficients)
```

```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

# -----------------------------
# Ridge Regression
# -----------------------------

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(X_train, y_train)

ridge_pred = ridge_model.predict(X_test)

ridge_mse = mean_squared_error(y_test, ridge_pred)
ridge_rmse = np.sqrt(ridge_mse)
ridge_r2 = r2_score(y_test, ridge_pred)


# -----------------------------
# Lasso Regression
# -----------------------------

lasso_model = Lasso(alpha=0.01)

lasso_model.fit(X_train, y_train)

lasso_pred = lasso_model.predict(X_test)

lasso_mse = mean_squared_error(y_test, lasso_pred)
lasso_rmse = np.sqrt(lasso_mse)
lasso_r2 = r2_score(y_test, lasso_pred)


# -----------------------------
# Model Comparison
# -----------------------------

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression"
    ],
    "MSE": [
        mse,
        ridge_mse,
        lasso_mse
    ],
    "RMSE": [
        rmse,
        ridge_rmse,
        lasso_rmse
    ],
    "R2 Score": [
        r2,
        ridge_r2,
        lasso_r2
    ]
})

display(comparison)
```
